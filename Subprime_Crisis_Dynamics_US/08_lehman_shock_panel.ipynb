```python
"""
Objective: Visually inspect the macroeconomic vector for the Rolling-Window VAR. 
Generates a stacked triple panel (TED Spread, Unemployment, and Fed Funds) highlighting 
the historical milestone of the Lehman Brothers collapse.
"""

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

import warnings
warnings.filterwarnings('ignore')

# Clean style configuration for academic papers
sns.set_theme(style="whitegrid", rc={"axes.spines.right": False, "axes.spines.top": False})

# ==========================================
# 1. DATA LOADING (MACRO VECTOR)
# ==========================================
def load_panel_data(csv_file: str) -> pd.DataFrame:
    """Reads the macroeconomic dataset generated in Step 07."""
    print(f"[*] Loading lab dataset ({csv_file})...")
    if not os.path.exists(csv_file):
        print(f"[ERROR] File '{csv_file}' not found. Run Step 07 first.")
        sys.exit(1)
        
    df = pd.read_csv(csv_file, parse_dates=['Date'], index_col='Date')
    return df

# ==========================================
# 2. TRIPLE PANEL RENDERING
# ==========================================
def render_shock_panel(df: pd.DataFrame, output_file: str):
    """Creates a stacked chart with the 3 variables and the Lehman Brothers line."""
    print("[*] Rendering the macroeconomic shock panel...")

    fig, axes = plt.subplots(3, 1, figsize=(10, 10), sharex=True)

    # Lehman Brothers collapse date (September 2008)
    lehman_date = pd.to_datetime('2008-09-15')

    # 1. TED Spread (Interbank Market Panic)
    axes[0].plot(df.index, df['TED_Spread'], color='firebrick', linewidth=2)
    axes[0].set_ylabel('TED Spread (%)', fontsize=10, fontweight='bold')
    axes[0].axvline(lehman_date, color='black', linestyle=':', linewidth=1.5, alpha=0.7)
    axes[0].text(lehman_date, df['TED_Spread'].max(), ' Lehman Bros. Collapse', 
                 verticalalignment='top', fontsize=9, fontweight='bold')

    # 2. Unemployment (Real Economy Contagion)
    axes[1].plot(df.index, df['Unemployment'], color='steelblue', linewidth=2)
    axes[1].set_ylabel('Unemployment (%)', fontsize=10, fontweight='bold')
    axes[1].axvline(lehman_date, color='black', linestyle=':', linewidth=1.5, alpha=0.7)

    # 3. Fed Funds Rate (Institutional/Central Bank Response)
    axes[2].plot(df.index, df['Fed_Funds'], color='forestgreen', linewidth=2)
    axes[2].set_ylabel('Fed Funds (%)', fontsize=10, fontweight='bold')
    axes[2].axvline(lehman_date, color='black', linestyle=':', linewidth=1.5, alpha=0.7)

    # Final adjustments (Title removed for native LaTeX insertion)
    plt.xlabel('Year', fontsize=11, fontweight='bold')
    plt.tight_layout()

    plt.savefig(output_file, dpi=300, bbox_inches='tight')
    plt.close()
    
    print(f"[SUCCESS] High-resolution chart exported: '{output_file}'")

if __name__ == "__main__":
    df_macro = load_panel_data('07_clean_rolling_var_fred.csv')
    render_shock_panel(df_macro, '08_lehman_shock_panel.png')
```
