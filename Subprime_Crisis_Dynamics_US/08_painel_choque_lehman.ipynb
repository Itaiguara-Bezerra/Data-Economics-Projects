```python
"""
Objetivo: Inspecionar visualmente o vetor macroeconômico do Rolling-Window VAR. 
Gera um painel triplo empilhado (TED Spread, Desemprego e Fed Funds) destacando 
o marco histórico do colapso do Lehman Brothers.
"""

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os
import warnings

warnings.filterwarnings('ignore')

# Configuração de estilo limpo para artigos acadêmicos
sns.set_theme(style="whitegrid", rc={"axes.spines.right": False, "axes.spines.top": False})

# ==========================================
# 1. CARREGAMENTO DOS DADOS (VETOR MACRO)
# ==========================================
def carregar_dados_painel(arquivo_csv: str) -> pd.DataFrame:
    """Lê a base macroeconômica gerada na Etapa 07."""
    print(f"[*] Carregando a base do laboratório ({arquivo_csv})...")
    if not os.path.exists(arquivo_csv):
        print(f"[ERRO] Arquivo '{arquivo_csv}' não encontrado. Rode a Etapa 07 primeiro.")
        sys.exit(1)
        
    df = pd.read_csv(arquivo_csv, parse_dates=['Data'], index_col='Data')
    return df

# ==========================================
# 2. RENDERIZAÇÃO DO PAINEL TRIPLO
# ==========================================
def renderizar_painel_choque(df: pd.DataFrame, arquivo_saida: str):
    """Cria um gráfico empilhado com as 3 variáveis e a linha do Lehman Brothers."""
    print("[*] Renderizando o painel de choque macroeconômico...")

    fig, axes = plt.subplots(3, 1, figsize=(10, 10), sharex=True)

    # Data do colapso do Lehman Brothers (Setembro de 2008)
    data_lehman = pd.to_datetime('2008-09-15')

    # 1. TED Spread (Pânico no Mercado Interbancário)
    axes[0].plot(df.index, df['TED_Spread'], color='firebrick', linewidth=2)
    axes[0].set_ylabel('TED Spread (%)', fontsize=10, fontweight='bold')
    axes[0].axvline(data_lehman, color='black', linestyle=':', linewidth=1.5, alpha=0.7)
    axes[0].text(data_lehman, df['TED_Spread'].max(), ' Quebra do Lehman Bros.', 
                 verticalalignment='top', fontsize=9, fontweight='bold')

    # 2. Desemprego (Contágio na Economia Real)
    axes[1].plot(df.index, df['Desemprego'], color='steelblue', linewidth=2)
    axes[1].set_ylabel('Desemprego (%)', fontsize=10, fontweight='bold')
    axes[1].axvline(data_lehman, color='black', linestyle=':', linewidth=1.5, alpha=0.7)

    # 3. Fed Funds Rate (Resposta Institucional/Banco Central)
    axes[2].plot(df.index, df['Fed_Funds'], color='forestgreen', linewidth=2)
    axes[2].set_ylabel('Fed Funds (%)', fontsize=10, fontweight='bold')
    axes[2].axvline(data_lehman, color='black', linestyle=':', linewidth=1.5, alpha=0.7)

    # Ajustes finais (Título removido para inserção nativa no LaTeX)
    plt.xlabel('Ano', fontsize=11, fontweight='bold')
    plt.tight_layout()

    plt.savefig(arquivo_saida, dpi=300, bbox_inches='tight')
    plt.close()
    
    print(f"[SUCESSO] Gráfico exportado com alta resolução: '{arquivo_saida}'")

if __name__ == "__main__":
    df_macro = carregar_dados_painel('07_base_rolling_var_fred.csv')
    renderizar_painel_choque(df_macro, '08_painel_choque_lehman.png')
```
